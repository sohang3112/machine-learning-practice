### 1. **Core Transformations in 2D**

* **Translation** – moves a point/object.
* **Scaling** – resizes.
* **Rotation** – spins around a pivot.

### 2. **Other Common 2D Operations**

Beyond these, the following are also widely used:

* **Reflection (Mirror)** - Flips an object about an axis (x-axis, y-axis, or arbitrary line).
* **Shearing (Skewing)** - Shifts layers of an object to "tilt" it. Example: turning a rectangle into a parallelogram.
* **Projection** - Maps 3D to 2D points. E.g., perspective or orthographic projection in rendering pipelines.
* **Homothety (Uniform scaling from a point)**: Special case of scaling, but explicitly about scaling relative to a specific point (not just the origin).
* **Composition of Transforms**: Combining multiple affine transformations into a single operation (matrix multiplication).

### 3. **Advanced (Beyond Affine)**

* **Non-linear Warping** (bending, fisheye, etc.).
* **Morphing** (smooth transition between shapes).
* **Conformal / Similarity transformations** (preserve angles but not necessarily lengths).
